```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity
from math import radians, cos, sin, asin, sqrt


class HybridCollegeMatcher:
    """
    Hybrid recommender:
      - PCA + cosine similarity on curated numeric features
      - PLUS explicit subscores for affordability, academics, fit, diversity/mobility, distance
      - Final score is a weighted combination using student preferences.
    """

    def __init__(self, curated_features, weights=None, n_components=20,
                 max_distance_km=3500):
        self.curated_features = curated_features
        self.weights = weights if weights else {}
        self.n_components = n_components
        self.max_distance_km = max_distance_km

        self.scaler = None
        self.pca = None
        self.numeric_features = None
        self.college_embeddings = None
        self.df_full = None
        self.medians = None

    # ------------------------------------------------------------
    # Haversine distance
    # ------------------------------------------------------------
    def haversine(self, lat1, lon1, lat2, lon2):
        lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
        dlon = lon2 - lon1
        dlat = lat2 - lat1
        a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
        c = 2 * asin(sqrt(a))
        return 6371 * c  # km

    # ------------------------------------------------------------
    # Feature Engineering
    # ------------------------------------------------------------
    def add_engineered_features(self, df):
        # Diversity entropy
        race_cols = [
            "Percent of White Undergraduates",
            "Percent of Black or African American Undergraduates",
            "Percent of Latino Undergraduates",
            "Percent of Asian Undergraduates",
            "Percent of American Indian or Alaska Native Undergraduates",
            "Percent of Native Hawaiian or Other Pacific Islander Undergraduates",
            "Percent of Two or More Races Undergraduates",
        ]
        race_cols = [c for c in race_cols if c in df.columns]
        if race_cols:
            race_props = df[race_cols].div(100)

            def entropy(row):
                p = row[row > 0]
                return -(p * np.log(p)).sum() if len(p) > 0 else 0.0

            df["diversity_entropy"] = race_props.apply(entropy, axis=1)
        else:
            df["diversity_entropy"] = 0.0

        # Degree shares (if not already precomputed)
        if "Number of Bachelor Degrees Grand Total" in df.columns:
            total = df["Number of Bachelor Degrees Grand Total"].replace(0, np.nan)

            share_map = {
                "Number of Degrees Awarded in Science, Technology, Engineering, and Math": "stem_share",
                "Number of Degrees Awarded in Business": "business_share",
                "Number of Degrees Awarded in Health Sciences": "health_share",
                "Number of Degrees Awarded in Arts and Humanities": "arts_share",
                "Number of Degrees Awarded in Education": "education_share",
                "Number of Degrees Awarded in Social Sciences": "soc_science_share",
            }
            for src, tgt in share_map.items():
                if src in df.columns and tgt not in df.columns:
                    df[tgt] = (df[src] / total).fillna(0)
        else:
            # Ensure columns exist
            for c in ["stem_share", "business_share", "health_share",
                      "arts_share", "education_share", "soc_science_share"]:
                if c not in df.columns:
                    df[c] = 0.0

        df = df.fillna(0)
        return df

    # ------------------------------------------------------------
    # Fit on colleges (no student-specific bits yet)
    # ------------------------------------------------------------
    def fit(self, df_full):
        df_full = df_full.copy()
        df_full = self.add_engineered_features(df_full)

        # Keep for later use
        self.df_full = df_full

        # Select curated numeric features
        FEATURES = [c for c in self.curated_features if c in df_full.columns]
        numeric_cols = df_full.select_dtypes(include=[np.number]).columns
        self.numeric_features = [c for c in FEATURES if c in numeric_cols]

        # Freeze medians for embedding construction
        self.medians = df_full[self.numeric_features].median()

        # Numeric matrix
        X = df_full[self.numeric_features].fillna(self.medians)

        # Apply feature weights BEFORE scaling
        for col, w in self.weights.items():
            if col in X.columns:
                X[col] *= w

        # Fit scaler + PCA
        self.scaler = StandardScaler()
        X_scaled = self.scaler.fit_transform(X)

        self.pca = PCA(n_components=min(self.n_components, len(self.numeric_features)))
        self.college_embeddings = self.pca.fit_transform(X_scaled)

        # store embeddings in df_full
        self.df_full["embedding"] = list(self.college_embeddings)

        return self

    # ------------------------------------------------------------
    # Student preference weights (NOT a fake college row yet)
    # ------------------------------------------------------------
    def _build_student_prefs(self, student):
        # default importance weights if not provided
        prefs = {
            "importance_afford": student.get("importance_afford", 0.4),
            "importance_outcomes": student.get("importance_outcomes", 0.3),
            "importance_diversity": student.get("importance_diversity", 0.2),
            "importance_mobility": student.get("importance_mobility", 0.2),
            "importance_location": student.get("importance_location", 0.2),
        }

        # Normalize so they sum to <= 1, keep a chunk reserved for similarity
        total = sum(prefs.values())
        if total > 0:
            for k in prefs:
                prefs[k] /= total

        # We'll give some fixed mass to PCA similarity
        prefs["importance_similarity"] = 0.2
        # Rescale the others to sum to 0.8
        non_sim_total = sum(v for k, v in prefs.items() if k != "importance_similarity")
        if non_sim_total > 0:
            for k in prefs:
                if k != "importance_similarity":
                    prefs[k] = prefs[k] * 0.8 / non_sim_total

        return prefs

    # ------------------------------------------------------------
    # Student-dependent per-college distance (km)
    # ------------------------------------------------------------
    def _compute_distances(self, student, df):
        home_lat = student.get("home_lat")
        home_lon = student.get("home_lon")

        if home_lat is None or home_lon is None:
            return pd.Series(np.nan, index=df.index)

        lats = df["Latitude"]
        lons = df["Longitude"]
        return pd.Series(
            [
                self.haversine(home_lat, home_lon, c_lat, c_lon)
                for c_lat, c_lon in zip(lats, lons)
            ],
            index=df.index
        )

    # ------------------------------------------------------------
    # Affordability score (0-100)
    # ------------------------------------------------------------
    def _score_affordability(self, student, df):
        target_price = student.get("target_net_price", 15000)
        will_work = student.get("will_work_job", True)
        max_hours = student.get("max_work_hours", 10)

        # Basic price
        if "Net Price" in df.columns:
            price = df["Net Price"].clip(lower=0)
        else:
            # fallback: in-state if exists
            if "Cost of Attendance: In State" in df.columns:
                price = df["Cost of Attendance: In State"]
            else:
                price = pd.Series(target_price, index=df.index)

        ratio = price / max(target_price, 1)  # price relative to target

        # Score: 100 if price <= target, linearly down to 0 at 2 * target, then clipped
        raw = 2.0 - ratio
        price_score = np.clip(raw, 0, 1) * 100

        # Work hours penalty (if data exists)
        hrs_cols = [
            "Weekly Hours to Close Gap",
            "Weekly Hours to Close Gap: Center-Based Care",
            "Weekly Hours to Close Gap: Home-Based Care",
        ]
        hrs_cols = [c for c in hrs_cols if c in df.columns]
        if hrs_cols and will_work:
            hrs = df[hrs_cols[0]].fillna(max_hours)
            # 0 hours -> 100, max_hours -> ~60, 2*max_hours -> 0
            norm = hrs / max(max_hours, 1)
            raw_hrs = 2.0 - norm
            hrs_score = np.clip(raw_hrs, 0, 1) * 100
            afford = 0.7 * price_score + 0.3 * hrs_score
        else:
            afford = price_score

        return afford

    # ------------------------------------------------------------
    # Distance score (0-100)
    # ------------------------------------------------------------
    def _score_distance(self, distances_km):
        # 0 distance -> 100
        # max_distance_km -> 0
        maxd = self.max_distance_km
        score = (1 - distances_km / maxd) * 100
        score = score.clip(lower=0, upper=100)
        # If distance is NaN (no home location), treat as neutral 70
        score = score.fillna(70)
        return score

    # ------------------------------------------------------------
    # Academic score (0-100)
    # ------------------------------------------------------------
    def _score_academics(self, student, df):
        sat = student.get("sat_score", None)

        grad_cols = [
            "Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total",
            "Bachelor's Degree Graduation Rate Within 4 Years - Total",
        ]
        grad_cols = [c for c in grad_cols if c in df.columns]

        if grad_cols:
            grad = df[grad_cols[0]].fillna(0) / 100.0  # 0–1
        else:
            grad = pd.Series(0.5, index=df.index)

        if "First-Time, Full-Time Retention Rate" in df.columns:
            retain = df["First-Time, Full-Time Retention Rate"].fillna(0) / 100.0
        else:
            retain = pd.Series(0.5, index=df.index)

        # Simple SAT alignment if SAT columns exist and student has a score
        if sat is not None and \
           "SAT Evidence Based Reading and Writing - 25th Percentile Score" in df.columns and \
           "SAT Evidence Based Reading and Writing - 75th Percentile Score" in df.columns:

            sat25 = df["SAT Evidence Based Reading and Writing - 25th Percentile Score"].fillna(0)
            sat75 = df["SAT Evidence Based Reading and Writing - 75th Percentile Score"].fillna(0)

            # If student's SAT is within [25, 75] → best alignment
            # Below 25th → decreasing alignment
            # Above 75th → still good, but slightly less optimal (may be overmatch / underfit)
            alignment = []
            for s25, s75 in zip(sat25, sat75):
                if s25 <= 0 or s75 <= 0:
                    alignment.append(0.6)  # unknown
                else:
                    if sat < s25:
                        alignment.append(max(0.0, sat / s25 * 0.8))
                    elif sat > s75:
                        # still decent but not perfect
                        alignment.append(0.9)
                    else:
                        alignment.append(1.0)
            align_series = pd.Series(alignment, index=df.index)
        else:
            align_series = pd.Series(0.8, index=df.index)

        # Weighted combination
        acad = (0.5 * grad + 0.3 * retain + 0.2 * align_series) * 100
        acad = acad.clip(0, 100)
        return acad

    # ------------------------------------------------------------
    # Diversity + mobility score (0-100)
    # ------------------------------------------------------------
    def _score_diversity_mobility(self, student, df):
        imp_div = student.get("importance_diversity", 0.0)
        imp_mob = student.get("importance_mobility", 0.0)
        self_eth = (student.get("self_ethnicity", "") or "").lower()

        # Diversity entropy normalized
        if "diversity_entropy" in df.columns:
            div = df["diversity_entropy"]
            max_div = div.max() if div.max() > 0 else 1.0
            div_norm = (div / max_div).clip(0, 1)
        else:
            div_norm = pd.Series(0.5, index=df.index)

        # Pell % as mobility proxy
        if "Percent of First-Time, Full-Time Undergraduates Awarded Pell Grants" in df.columns:
            pell = df["Percent of First-Time, Full-Time Undergraduates Awarded Pell Grants"].fillna(0) / 100.0
        else:
            pell = pd.Series(0.5, index=df.index)

        # MSI boost if aligned with ethnicity
        msi_boost = pd.Series(0.0, index=df.index)
        if self_eth:
            if "latino" in self_eth or "hispanic" in self_eth:
                if "HSI" in df.columns:
                    msi_boost += df["HSI"].fillna(0) * 0.2
            if "black" in self_eth or "african" in self_eth:
                if "HBCU" in df.columns:
                    msi_boost += df["HBCU"].fillna(0) * 0.25
                if "PBI" in df.columns:
                    msi_boost += df["PBI"].fillna(0) * 0.15
            if "asian" in self_eth:
                if "AANAPII" in df.columns:
                    msi_boost += df["AANAPII"].fillna(0) * 0.2
            # etc for other categories if you want

        div_score = div_norm * 100
        mob_score = pell * 100

        score = (imp_div * div_score + imp_mob * mob_score) / max(imp_div + imp_mob, 1e-6)
        score = score + msi_boost * 100
        score = score.clip(0, 100)
        return score

    # ------------------------------------------------------------
    # Fit score (major, size, sector) (0-100)
    # ------------------------------------------------------------
    def _score_fit(self, student, df):
        intended = (student.get("intended_major", "") or "").lower()
        major_map = {
            "stem": "stem_share",
            "science": "stem_share",
            "engineering": "stem_share",
            "business": "business_share",
            "health": "health_share",
            "nursing": "health_share",
            "arts": "arts_share",
            "humanities": "arts_share",
            "education": "education_share",
            "social": "soc_science_share",
            "social science": "soc_science_share",
        }

        if intended in major_map and major_map[intended] in df.columns:
            major_fit = df[major_map[intended]].fillna(0)  # already 0–1-ish
        else:
            # if they just said "STEM"
            if intended == "stem" and "stem_share" in df.columns:
                major_fit = df["stem_share"].fillna(0)
            else:
                major_fit = pd.Series(0.5, index=df.index)

        # Size preference (Institution Size Category: 1,2,3 small/med/large)
        size_pref = student.get("pref_size", None)
        size_map = {"small": 1, "medium": 2, "large": 3}
        if isinstance(size_pref, str):
            size_pref_num = size_map.get(size_pref.lower(), None)
        else:
            size_pref_num = size_pref

        if "Institution Size Category" in df.columns and size_pref_num is not None:
            size_diff = (df["Institution Size Category"] - size_pref_num).abs()
            # 0 diff -> 1, 1 diff -> 0.7, 2 diff -> 0.4, etc.
            size_score = np.clip(1.0 - 0.3 * size_diff, 0.0, 1.0)
        else:
            size_score = pd.Series(0.7, index=df.index)

        # Sector preference (public/private)
        sector_pref = (student.get("pref_sector", "") or "").lower()
        sector_map = {"public": 1, "private": 2}  # adjust if your coding is different
        sector_pref_num = sector_map.get(sector_pref, None)

        if "Control of Institution" in df.columns and sector_pref_num is not None:
            sec_diff = (df["Control of Institution"] - sector_pref_num).abs()
            sec_score = np.clip(1.0 - 0.5 * sec_diff, 0.0, 1.0)
        else:
            sec_score = pd.Series(0.7, index=df.index)

        fit = (0.5 * major_fit + 0.25 * size_score + 0.25 * sec_score) * 100
        fit = fit.clip(0, 100)
        return fit

    # ------------------------------------------------------------
    # PCA similarity score (0-100)
    # ------------------------------------------------------------
    def _score_similarity(self, student):
        # Build a "target school" vector using medians + preference nudges
        row = self.medians.copy()

        target_price = student.get("target_net_price", None)
        if target_price is not None and "Net Price" in row:
            row["Net Price"] = target_price

        intended = (student.get("intended_major", "") or "").lower()
        major_map = {
            "stem": "stem_share",
            "business": "business_share",
            "health": "health_share",
            "arts": "arts_share",
            "education": "education_share",
            "social": "soc_science_share",
        }
        if intended in major_map and major_map[intended] in row:
            # Nudge that share towards upper quartile
            # (approx: 2x median but clipped to max)
            col = major_map[intended]
            # you could also use quantiles from df_full if you want
            row[col] = row[col] * 2.0

        # Prefer higher grad rates / outcomes if they care about outcomes
        imp_out = student.get("importance_outcomes", 0.0)
        if imp_out > 0:
            for c in [
                "Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total",
                "First-Time, Full-Time Retention Rate",
                "Median Earnings of Students Working and Not Enrolled 10 Years After Entry",
            ]:
                if c in row.index:
                    row[c] = row[c] * (1.0 + 0.5 * imp_out)

        # Convert to DataFrame
        student_vec = pd.DataFrame([row[self.numeric_features]])

        # Apply same weights as colleges
        for col, w in self.weights.items():
            if col in student_vec.columns:
                student_vec[col] *= w

        # Scale + PCA transform
        student_scaled = self.scaler.transform(student_vec)
        student_emb = self.pca.transform(student_scaled)

        sims = cosine_similarity(self.college_embeddings, student_emb).flatten()
        # convert [-1,1] -> [0,100]
        sim_score = ((sims + 1) / 2) * 100
        return sim_score

    # ------------------------------------------------------------
    # Main match function
    # ------------------------------------------------------------
    def match(self, student, top_k=10, return_components=False):
        """
        student: dict with keys like:
            - target_net_price, max_work_hours, will_work_job
            - sat_score, act_score
            - importance_afford, importance_outcomes, importance_diversity,
              importance_mobility, importance_location
            - pref_size, pref_sector
            - home_lat, home_lon
            - intended_major, self_ethnicity
        """
        if self.df_full is None:
            raise ValueError("Call fit(df_full) before match().")

        prefs = self._build_student_prefs(student)

        # Work on a copy so we don't mutate the training df
        df = self.df_full.copy()

        # ------------------------------------------------------------
        # HARD FILTERS (must apply before ANY scoring)
        # ------------------------------------------------------------
        
        # 1. Must award Bachelor’s degrees (remove 2-year/trade schools)
        if "Number of Bachelor Degrees Grand Total" in df.columns:
            df = df[df["Number of Bachelor Degrees Grand Total"] > 0]
        
        # 2. Remove trade/vocational/cosmetology/beauty/barber institutes
        INVALID = [
            "Cosmetology", "Beauty", "Barber", "Massage", "Therapy",
            "Academy", "Institute", "Technical", "Vocational", "Career"
        ]
        df = df[~df["Institution Name"].str.contains('|'.join(INVALID), case=False, na=False)]
        
        # 3. Remove for-profits (Control 3)
        if "Control of Institution" in df.columns:
            df = df[df["Control of Institution"] != 3]
        
        # 4. Minimum academic quality
        if "First-Time, Full-Time Retention Rate" in df.columns:
            df = df[df["First-Time, Full-Time Retention Rate"] > 55]
        
        if "Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total" in df.columns:
            df = df[df["Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total"] > 30]
        
        # 5. Major-specific filters (example: STEM)
        if student.get("intended_major", "").lower() == "stem":
            if "stem_share" in df.columns:
                df = df[df["stem_share"] > 0.05]

        # Distance
        dist_km = self._compute_distances(student, df)
        df["distance_km"] = dist_km

        # Hard distance cutoff
        mask_near = dist_km.isna() | (dist_km <= self.max_distance_km)
        df = df[mask_near].copy()

        # Reindex distance_km
        dist_km = df["distance_km"]

        # Subscores
        afford_score = self._score_affordability(student, df)
        acad_score = self._score_academics(student, df)
        divmob_score = self._score_diversity_mobility(student, df)
        fit_score = self._score_fit(student, df)
        dist_score = self._score_distance(dist_km)
        sim_score_full = self._score_similarity(student)

        # similarity scores need to be aligned with df index (same order as original df_full)
        sim_score = pd.Series(sim_score_full, index=self.df_full.index).loc[df.index]

        # Final weighted score
        final = (
            prefs["importance_afford"] * afford_score +
            prefs["importance_outcomes"] * acad_score +
            prefs["importance_diversity"] * divmob_score +
            prefs["importance_mobility"] * divmob_score +  # mobility piggybacks on same score
            prefs["importance_location"] * dist_score +
            prefs["importance_similarity"] * sim_score
        )

        df["match_score"] = final.clip(0, 100)

        # Sort and dedupe by institution name
        inst_col = "Institution Name" if "Institution Name" in df.columns else "Institution Name_x"
        df = df.sort_values("match_score", ascending=False)
        df = df.drop_duplicates(subset=[inst_col], keep="first")

        base_cols = [inst_col, "match_score", "distance_km"]
        extra_cols = [
            "Net Price",
            "Cost of Attendance: In State",
            "Cost of Attendance: Out of State",
        ]
        extra_cols = [c for c in extra_cols if c in df.columns]

        if return_components:
            comp_cols = {
                "afford_score": afford_score,
                "academic_score": acad_score,
                "diversity_mobility_score": divmob_score,
                "fit_score": fit_score,
                "distance_score": dist_score,
                "similarity_score": sim_score,
            }
            for name, series in comp_cols.items():
                df[name] = series

            cols = base_cols + extra_cols + list(comp_cols.keys())
            cols = [c for c in cols if c in df.columns]
            return df[cols].head(top_k)

        else:
            cols = base_cols + extra_cols
            cols = [c for c in cols if c in df.columns]
            return df[cols].head(top_k)
```

```python
# Load the affordability dataset
df_afford = pd.read_excel("data/Affordability Gap Data AY2022-23 2.17.25.xlsx")

# Load the college results dataset
df_results = pd.read_excel("data/College Results View 2021 Data Dump for Export.xlsx")

df = df_afford.merge(
    df_results,
    left_on="Unit ID",
    right_on="UNIQUE_IDENTIFICATION_NUMBER_OF_THE_INSTITUTION",
    how="inner"
)

total = df["Number of Bachelor Degrees Grand Total"].replace(0, np.nan)

engineered = pd.DataFrame({
    "stem_share":        df["Number of Degrees Awarded in Science, Technology, Engineering, and Math"] / total,
    "business_share":    df["Number of Degrees Awarded in Business"] / total,
    "health_share":      df["Number of Degrees Awarded in Health Sciences"] / total,
    "arts_share":        df["Number of Degrees Awarded in Arts and Humanities"] / total,
    "education_share":   df["Number of Degrees Awarded in Education"] / total,
    "soc_science_share": df["Number of Degrees Awarded in Social Sciences"] / total
}).fillna(0)

df = pd.concat([df, engineered], axis=1)

CURATED_FEATURES = (
    afford_cols
    + academic_cols
    + economic_cols
    + demo_cols
    + degree_cols
    + location_cols
    + msi_cols
    + [
        "stem_share",
        "business_share",
        "health_share",
        "arts_share",
        "education_share",
        "soc_science_share"
    ]
)

FEATURES = [c for c in CURATED_FEATURES if c in df.columns]

df = df.rename(columns={"Institution Name_x": "Institution Name"})
df = df.drop(columns=["Institution Name_y"], errors="ignore")

for col in df.columns:
    if df[col].isna().all():
        df[col] = 0

# Fill numeric columns with medians
for col in df.select_dtypes(include=[np.number]).columns:
    df[col] = df[col].fillna(df[col].median())

df_full = df.copy()

num_cols = df_full.select_dtypes(include=[np.number]).columns
numeric_features = [c for c in FEATURES if c in num_cols]

X = df_full[numeric_features]
```

```python

matcher = HybridCollegeMatcher(
    curated_features=CURATED_FEATURES,
    weights=weights,
    n_components=20,
    max_distance_km=3500  # or shrink this if you want more local results
)
matcher.fit(df_full)
```
